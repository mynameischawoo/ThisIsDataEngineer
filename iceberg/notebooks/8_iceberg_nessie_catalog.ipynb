# 학습 목표

# TL;DR
- Nessie는 데이터 레이크에 Git의 버전 관리 개념을 도입한 카탈로그로, Iceberg 데이터의 변화를 안전하게 추적하고 실험할 수 있게 해준다.
| 항목         | 요약                                        |
| ---------- | ----------------------------------------- |
| 핵심 개념      | Nessie = Git for Data Lakes               |
| 역할         | 데이터는 S3에 그대로 두고, **변경 이력(메타데이터)** 만 버전 관리 |
| 기능         | Branch, Commit, Merge, Tag 제공             |
| 장점         | 안전한 실험, 버전 관리, 협업, 복구 가능                  |
| Iceberg 연동 | Iceberg의 메타데이터 변경을 Nessie가 추적 및 관리        |
